```python
# Bank Management System using Python Functions
     
print(">>>>>>>>>>>>>>>>>>>>>>>>>>>>>>>>")
print("   WELCOME TO OUR BANK SYSTEM   ")
print("<<<<<<<<<<<<<<<<<<<<<<<<<<<<<<<<")
print("\n1. Creating Account")
print("2. Deposite Amount")
print("3. Withdraw Amount")
print("4. Check Balance")
print("5. Account Details")
print("7. Exist")

accounts = []

def creating_account():
    print("\n>>>>>>CREATING ACCOUNT<<<<<<")
    account_number = input("Enter your account number :")
    account_name = input("Enter your name :")
    initial_balance = float(input("Enter your initial balance :"))
    account_type = input("Enter your account type (savings/current) choice one :")
    for account in accounts:
        if account["account_number"] == account_number:
            print("Account number already Exist")
            return
    if initial_balance < 0:
            print("Initial Balance cannot be negative")
            return

    account={
            "account_number": account_number,
            "account_name"  : account_name,
            "balance"       : initial_balance,
            "account_type"  : account_type       
            }
    accounts.append(account)
    print(f"Account_Number : {account_number}"),
    print(f"Account_Name   : {account_name}"),
    print(f"Balance        : {initial_balance}"),
    print(f"Account_Type   : {account_type}")
    print(">>>>> Account created Successfully <<<<<")

def deposite_amount():
    print("\n>>>>>> AMOUNT DEPOSITING <<<<<<")
    account_number =input("Enter your account number")
    amount =float(input("Enter your Deposite amount"))

    for account in accounts: 
            if account["account_number"] == account_number:
                if amount > 0:
                    account["balance"] += amount
                    print(f"Account Number : {account_number}")
                    print(f"Depositing amount : {amount}")
                    print(f"Total_Balance     :{account["balance"]}")
                    print(">>>>> Amount Deposited successfully <<<<<")

                else:
                    print("Amount must be grater than zero")
                    return
    else:
        print("Account number incorrect")
                

def withdraw_amount():
    print("\n>>>>> Withdrawing Amount <<<<<")
    account_number =input("Enter your account number")
    amount =float(input("Enter a amount"))

    for account in accounts:
        if account["account_number"] == account_number:
            if amount > account["balance"]:
                print("Withdrawal Amount Grater Than Balance")
                return
            elif amount <= 0:
                print("Withdraw Amount Not in Negative")
                return
            else:
                account["balance"] -= amount
                print(f"Account Number : {account_number}")
                print(f"withdraw Amount : {amount}")
                print(f"Balance         : {account["balance"]}")
                print(">>>>> Amount Withdrawed successfully <<<<<")
    else:
        print("Account Number is Incorrect")
            

def check_balance():
    print("\n>>>>> Checking Account Balance <<<<<")
    account_number=input("Enter your account number")
    for account in accounts:
        if account["account_number"] == account_number:
            print(f"Account Balance : {account["balance"]}")
        else:
            print("Account Number is Incorrect")

def account_details():
    print("\n=======> ALL ACCOUNTS DETAILS <=======")
    if len(accounts) == 0:
        print("No Account Created")
        return
    for account in accounts:
        
            print(f"\nAccount_Number : {account["account_number"]}"),
            print(f"Account_Name   : {account["account_name"]}"),
            print(f"Balance        : {account["balance"]}"),
            print(f"Account_Type   : {account["account_type"]}")
            print("----------------------------------------------")
           
while True:

    choice =input("Enter your choice")

    if choice == "1":
        creating_account()
    elif choice == "2":
        deposite_amount()
    elif choice == "3":
        withdraw_amount()
    elif choice == "4":
        check_balance()
    elif choice == "5":
        account_details()
    elif choice == "7":
        print("\n>>>>>>>>>>>>>>>>>>>>>>>>>>>>>>>>>>>>>")
        print("THANK YOU FOR USEING OUR BANK SYSTEM")
        print("<<<<<<<<<<<<<<<<<<<<<<<<<<<<<<<<<<<<<")
        break
    else:
        print("Enter a Given Choice Number")
```
