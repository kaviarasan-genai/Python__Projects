```python
# Multiple linear regression
# Task 2 : employee salary predictor
data = {
    "Exprience":[1,2,3,5,7],
    "age":[22,24,26,29,32],
    "certification":[0,1,1,2,3],
    "salary":[25000,32000,38000,55000,72000]
}
df =pd.DataFrame(data)

x =df[["Exprience","age","certification"]].values
y =df["salary"].values

x_mean=np.mean(x,axis=0)
y_mean=np.mean(y)

x_cen = x-x_mean
y_cen = y-y_mean

m=np.linalg.inv(x_cen.T.dot(x_cen)).dot(x_cen.T).dot(y_cen)

b = y_mean -np.dot(x_mean ,m)

print("Slopes    :",m)
print("Intercept :",int(b))

new_x = np.array([6,30,2])

predic = np.dot(new_x,m)+b
print("Expected Salary :",int(predic))
```
