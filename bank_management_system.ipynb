```python
# Bank Management System using Python OOPS concepts
print("=========================================================")
print(" <.....B.A.N.K...M.A.N.A.G.E.M.E.N.T...S.Y.S.T.E.M.....>" )
print("=========================================================")
print("1. Creating account")
print("2. Deposite amount")
print("3. Withdraw amount")
print("4. Check Balance")
print("5. Account Details")
print("6. Exist")

class bank_system:
    def __init__(self,account_number,account_name,balance,account_type):
        self.account_number = account_number
        self.account_name = account_name
        self.balance = balance
        self.account_type = account_type

    def deposite_amount(self):
        amount = float(input("Enter deposite amount :"))
        if amount > 0:
            self.balance += amount
            print("\n==== Amount credited successfully...")
            print("Current Balance :",self.balance)
        else:
            print("\nAmount not deposited...")

    def withdraw(self):
        amount = float(input("Enter withdraw amount :"))
        if amount <= 0:
              print("\nAmount must be greater than Zero ")
        elif amount <= self.balance :
              self.balance -= amount
              print("\nAmount withdraw successfully")
              print("Current balance :",self.balance)
        else:
             print("\nInsufficient Balance")

    def check_balance(self):
         print("Current Balance :",self.balance)

    def account_details(self):
        print("----------ACCOUNT DETAILS----------")
        print("ACCOUNT NUMBER :",self.account_number)
        print("ACCOUNT NAME   :",self.account_name)
        print("BALANCE        :",self.balance)
        print("ACCOUNT TYPE   :",self.account_type)
        print("-------------------------------------")

accounts = []

def creating_account():
    account_number = input("Enter your Account Number")
    account_name = input("Enter your Name")
    balance = float(input("Enter your initial Balance"))
    account_type = input("Enter account type (savings/current)")
     
    account = bank_system(
    account_number,
    account_name,
    balance,
    account_type 
    )
    
    accounts.append(account)
    print("\n>>> Account created successfully <<<")
  
def find_account(account_number):
    for account in accounts:
        if account.account_number == account_number:
          return account
    
    return None

while True:

    choose = input("Enter Your Choice")

    if choose == "1":
        creating_account()

    elif choose == "2":
        account_number = input("Enter a account number")
        account = find_account(account_number)
        if account :
            account.deposite_amount()
        else:
            print("Account not Found ...")

    elif choose == "3":
        account_number = input("Enter a account number")
        account =find_account(account_number)
        if account:
            account.withdraw()
        else:
            print("Account not Found ...")
    elif choose == "4":
        account_number =input("Enter a account number")
        account =find_account(account_number)
        if account:
            account.check_balance()
        else:
            print("Account not Found ...")

    elif choose == "5":
        account_number =input("Enter a account number")
        account =find_account(account_number)
        if account:
            account.account_details()
        else:
            print("Account not Found")
    elif choose == "6":
        print("\n>>>>>>>>>>>>>>>>>>>>>>>>>>>>>>>>>>>>>>>>") 
        print("   | THANKING FOR USING BANK SYSTEM |"   )
        print(">>>>>>>>>>>>>>>>>>>>>>>>>>>>>>>>>>>>>>>>") 
        break

    else:
        print("Choose a given options ...")
```
